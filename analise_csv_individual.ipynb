## Função para Análise Individual de CSV

```python
def analisar_csv(file_path):
    """
    Função completa para análise individual de um arquivo CSV
    """
    file_name = os.path.basename(file_path).replace('.csv', '')
    
    # Carregar dados
    df = pd.read_csv(file_path)
    
    print("=" * 80)
    print(f"📄 ARQUIVO: {file_name}")
    print("=" * 80)
    
    # 1. INFORMAÇÕES BÁSICAS
    print(f"\n📊 INFORMAÇÕES BÁSICAS:")
    print(f"  • Dimensões: {df.shape[0]:,} linhas × {df.shape[1]} colunas")
    print(f"  • Tamanho em memória: {df.memory_usage(deep=True).sum() / 1024:.2f} KB")
    print(f"  • Tipos de dados: {df.dtypes.value_counts().to_dict()}")
    
    # 2. COLUNAS
    print(f"\n📋 COLUNAS ({df.shape[1]}):")
    for col in df.columns:
        dtype = df[col].dtype
        print(f"  • {col}: {dtype}")
    
    # 3. VALORES FALTANTES
    missing = df.isnull().sum()
    if missing.sum() > 0:
        print(f"\n⚠️ VALORES FALTANTES:")
        for col in missing[missing > 0].index:
            pct = (missing[col] / len(df)) * 100
            print(f"  • {col}: {missing[col]} ({pct:.2f}%)")
    else:
        print(f"\n✅ VALORES FALTANTES: Nenhum encontrado")
    
    # 4. ESTATÍSTICAS DESCRITIVAS
    numeric_cols = df.select_dtypes(include=[np.number]).columns
    if len(numeric_cols) > 0:
        print(f"\n📈 ESTATÍSTICAS DESCRITIVAS ({len(numeric_cols)} colunas numéricas):")
        stats_df = df[numeric_cols].describe().round(4)
        print(stats_df)
    else:
        print(f"\n📈 ESTATÍSTICAS: Sem colunas numéricas")
    
    # 5. DUPLICATAS
    duplicates = df.duplicated().sum()
    print(f"\n🔄 DUPLICATAS: {duplicates} linhas duplicadas")
    
    # 6. PRIMEIRAS E ÚLTIMAS LINHAS
    print(f"\n🔍 AMOSTRA DE DADOS (primeiras 3 linhas):")
    print(df.head(3).to_string())
    
    print(f"\n🔍 ÚLTIMAS 3 LINHAS:")
    print(df.tail(3).to_string())
    
    return df

# Usar a função em um arquivo de exemplo (8º arquivo)
if len(csv_files) >= 8:
    sample_df = analisar_csv(csv_files[7])
else:
    print("Número insuficiente de arquivos")
```

## Visualizações para Cada CSV

```python
def visualizar_csv(file_path):
    """
    Gera visualizações para um arquivo CSV
    """
    file_name = os.path.basename(file_path).replace('.csv', '')
    df = pd.read_csv(file_path)
    
    numeric_cols = df.select_dtypes(include=[np.number]).columns
    
    if len(numeric_cols) == 0:
        print(f"Sem dados numéricos em {file_name}")
        return
    
    # Criar figura com 4 subplots
    fig = plt.figure(figsize=(16, 12))
    fig.suptitle(f'Análise: {file_name}', fontsize=16, fontweight='bold', y=0.995)
    
    # Subplot 1: Série temporal (primeiras 3 colunas numéricas)
    ax1 = plt.subplot(2, 3, 1)
    cols_to_plot = numeric_cols[:3]
    for col in cols_to_plot:
        ax1.plot(df[col], label=col, linewidth=1.5, alpha=0.7)
    ax1.set_title('Série Temporal (Primeiras 3 colunas)', fontweight='bold')
    ax1.set_xlabel('Índice')
    ax1.set_ylabel('Valor')
    ax1.legend(fontsize=8)
    ax1.grid(True, alpha=0.3)
    
    # Subplot 2: Histograma da primeira coluna numérica
    ax2 = plt.subplot(2, 3, 2)
    first_col = numeric_cols[0]
    ax2.hist(df[first_col].dropna(), bins=40, color='skyblue', edgecolor='black', alpha=0.7)
    ax2.set_title(f'Distribuição: {first_col}', fontweight='bold')
    ax2.set_xlabel('Valor')
    ax2.set_ylabel('Frequência')
    ax2.grid(True, alpha=0.3, axis='y')
    
    # Subplot 3: Box plot
    ax3 = plt.subplot(2, 3, 3)
    data_to_box = [df[col].dropna() for col in cols_to_plot]
    bp = ax3.boxplot(data_to_box, labels=cols_to_plot, patch_artist=True)
    for patch in bp['boxes']:
        patch.set_facecolor('lightblue')
    ax3.set_title('Box Plot', fontweight='bold')
    ax3.set_ylabel('Valor')
    ax3.tick_params(axis='x', rotation=45, labelsize=8)
    ax3.grid(True, alpha=0.3, axis='y')
    
    # Subplot 4: Matriz de correlação (se houver mais de 1 coluna numérica)
    ax4 = plt.subplot(2, 3, 4)
    if len(numeric_cols) > 1:
        corr = df[numeric_cols[:5]].corr()
        sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0, 
                    ax=ax4, cbar_kws={'label': 'Correlação'}, linewidths=0.5)
        ax4.set_title('Matriz de Correlação', fontweight='bold')
    else:
        ax4.text(0.5, 0.5, 'Apenas 1 coluna numérica', ha='center', va='center')
        ax4.set_xticks([])
        ax4.set_yticks([])
    
    # Subplot 5: Desvio padrão móvel
    ax5 = plt.subplot(2, 3, 5)
    rolling_std = df[first_col].rolling(window=20).std()
    ax5.plot(rolling_std, color='purple', linewidth=2, label='Desvio Padrão Móvel (20)')
    ax5.fill_between(range(len(rolling_std)), rolling_std, alpha=0.3, color='purple')
    ax5.set_title('Variabilidade (Desvio Padrão Móvel)', fontweight='bold')
    ax5.set_xlabel('Índice')
    ax5.set_ylabel('Desvio Padrão')
    ax5.legend(fontsize=8)
    ax5.grid(True, alpha=0.3)
    
    # Subplot 6: Estatísticas resumidas em texto
    ax6 = plt.subplot(2, 3, 6)
    ax6.axis('off')
    
    stats_text = f"""
RESUMO ESTATÍSTICO

Dimensões: {df.shape[0]:,} x {df.shape[1]}
Colunas Numéricas: {len(numeric_cols)}

{first_col}:
  Média: {df[first_col].mean():.4f}
  Mediana: {df[first_col].median():.4f}
  Std Dev: {df[first_col].std():.4f}
  Min: {df[first_col].min():.4f}
  Max: {df[first_col].max():.4f}
  
Valores Faltantes: {df.isnull().sum().sum()}
Duplicatas: {df.duplicated().sum()}
"""
    
    ax6.text(0.1, 0.5, stats_text, fontsize=10, family='monospace',
             verticalalignment='center', bbox=dict(boxstyle='round', facecolor='wheat', alpha=0.5))
    
    plt.tight_layout()
    plt.show()

# Exemplo: visualizar o 8º arquivo
if len(csv_files) >= 8:
    visualizar_csv(csv_files[7])
```

## Analisar Todos os Arquivos CSV (8º a 18º)

```python
# Selecionar arquivos da 8ª até 18ª posição
selected_csv_files = csv_files[7:18]

print(f"\n{'='*80}")
print(f"ANALISANDO {len(selected_csv_files)} ARQUIVOS (8ª a 18ª posição)")
print(f"{'='*80}\n")

# Criar um dicionário para armazenar resumos
resumo_analise = {}

for idx, file_path in enumerate(selected_csv_files, 8):
    file_name = os.path.basename(file_path).replace('.csv', '')
    df = pd.read_csv(file_path)
    
    numeric_cols = df.select_dtypes(include=[np.number]).columns
    
    resumo_analise[file_name] = {
        'arquivo': idx,
        'linhas': df.shape[0],
        'colunas': df.shape[1],
        'colunas_numericas': len(numeric_cols),
        'valores_faltantes': df.isnull().sum().sum(),
        'duplicatas': df.duplicated().sum()
    }
    
    print(f"{idx}. {file_name}")
    print(f"   Tamanho: {df.shape[0]:,} linhas × {df.shape[1]} colunas")
    print(f"   Colunas numéricas: {len(numeric_cols)}")
    print(f"   Valores faltantes: {df.isnull().sum().sum()}")
    print()

# Exibir resumo em formato de tabela
print(f"\n{'='*80}")
print("RESUMO COMPARATIVO")
print(f"{'='*80}\n")

resumo_df = pd.DataFrame(resumo_analise).T
print(resumo_df.to_string())
```

## Análise de Anomalias e Outliers

```python
# Detectar anomalias no arquivo selecionado
print("=" * 80)
print("DETECÇÃO DE ANOMALIAS (Z-score > 3)")
print("=" * 80)

numeric_cols = df_selecionado.select_dtypes(include=[np.number]).columns

anomalias_encontradas = False

for col in numeric_cols:
    try:
        z_scores = np.abs(zscore(df_selecionado[col].dropna()))
        anomalies = np.where(z_scores > 3)[0]
        
        if len(anomalies) > 0:
            anomalias_encontradas = True
            print(f"\n{col}:")
            print(f"  • {len(anomalies)} anomalias detectadas ({(len(anomalies)/len(df_selecionado))*100:.2f}%)")
            print(f"  • Índices: {anomalies[:10].tolist()}{'...' if len(anomalies) > 10 else ''}")
            print(f"  • Valores anômalos: {df_selecionado[col].iloc[anomalies[:5]].values}")
    except Exception as e:
        pass

if not anomalias_encontradas:
    print("\n✅ Nenhuma anomalia significativa detectada (Z-score ≤ 3)")

# Visualizar distribuição e outliers
numeric_cols_list = list(numeric_cols)
if len(numeric_cols_list) > 0:
    fig, axes = plt.subplots(2, 2, figsize=(16, 10))
    fig.suptitle('Análise de Outliers e Distribuições', fontsize=14, fontweight='bold')
    
    for idx, col in enumerate(numeric_cols_list[:4]):
        ax = axes.flatten()[idx]
        
        # Box plot com violin plot
        parts = ax.violinplot([df_selecionado[col].dropna()], vert=True, widths=0.7,
                              showmeans=True, showmedians=True)
        
        # Adicionar scatter dos outliers
        Q1 = df_selecionado[col].quantile(0.25)
        Q3 = df_selecionado[col].quantile(0.75)
        IQR = Q3 - Q1
        outliers = df_selecionado[(df_selecionado[col] < Q1 - 1.5*IQR) | 
                                  (df_selecionado[col] > Q3 + 1.5*IQR)][col]
        
        if len(outliers) > 0:
            ax.scatter([1]*len(outliers), outliers, color='red', s=100, alpha=0.6, label='Outliers')
        
        ax.set_title(f'{col}\n({len(outliers)} outliers)', fontweight='bold')
        ax.set_ylabel('Valor')
        ax.grid(True, alpha=0.3, axis='y')
        if len(outliers) > 0:
            ax.legend()
    
    # Remover subplot vazio se houver menos de 4 colunas
    if len(numeric_cols_list) < 4:
        for idx in range(len(numeric_cols_list), 4):
            fig.delaxes(axes.flatten()[idx])
    
    plt.tight_layout()
    plt.show()
```
