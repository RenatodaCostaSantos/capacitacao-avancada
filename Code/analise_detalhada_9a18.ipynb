```python
def gerar_relatorio_detalhado(file_path, posicao):
    """
    Gera relatório detalhado e contextualizado para um arquivo CSV
    """
    file_name = os.path.basename(file_path).replace('carbonZ_2018-07-18-12-10-11_no_ground_truth-', '').replace('.csv', '')
    df = pd.read_csv(file_path)
    
    # Obter contexto
    ctx = contexto_arquivos.get(os.path.basename(file_path), {
        'tipo': 'Desconhecido',
        'origem': 'Desconhecido',
        'contexto': 'Desconhecido',
        'descricao': 'Desconhecido',
        'funcionalidades': []
    })
    
    # Cabeçalho
    print("\n" + "="*90)
    print(f"{'ARQUIVO: ' + file_name:^90}")
    print("="*90)
    
    # Visão geral
    print(f"\n📄 VISÃO GERAL DO ARQUIVO\n")
    print(f"Total de linhas: {df.shape[0]:,}")
    print(f"Total de colunas: {df.shape[1]}")
    print(f"Tipo de dado: {ctx['tipo']}")
    print(f"Origem: {ctx['origem']}")
    print(f"Contexto típico: {ctx['contexto']}")
    
    # Descrição
    print(f"\n{ctx['descricao']}")
    
    # Funcionalidades
    print(f"\nEm resumo: o que esse CSV te permite fazer\n")
    for func in ctx['funcionalidades']:
        print(f"{func}")
    
    # Informações técnicas
    numeric_cols = df.select_dtypes(include=[np.number]).columns
    print(f"\n\n📊 INFORMAÇÕES TÉCNICAS\n")
    print(f"Colunas numéricas: {len(numeric_cols)} de {df.shape[1]}")
    print(f"Tamanho em memória: {df.memory_usage(deep=True).sum() / 1024:.2f} KB")
    print(f"Valores faltantes: {df.isnull().sum().sum()}")
    print(f"Linhas duplicadas: {df.duplicated().sum()}")
    
    # Colunas
    print(f"\n📋 COLUNAS DISPONÍVEIS\n")
    for i, col in enumerate(df.columns, 1):
        dtype = df[col].dtype
        null_count = df[col].isnull().sum()
        print(f"{i:2d}. {col:<40} {str(dtype):<12} (faltantes: {null_count})")
    
    # Estatísticas descritivas
    if len(numeric_cols) > 0:
        print(f"\n📈 ESTATÍSTICAS DESCRITIVAS\n")
        stats_summary = df[numeric_cols].describe().round(6)
        print(stats_summary)
    
    # Amostra
    print(f"\n🔍 AMOSTRA DE DADOS (primeiras 3 linhas)\n")
    print(df.head(3).to_string())
    
    return df, ctx

# Analisar primeiro arquivo como exemplo
primeiro_arquivo = selected_files[0]
df_example, ctx_example = gerar_relatorio_detalhado(primeiro_arquivo, 9)
```

## Visualizações Completas para Cada Arquivo

```python
def gerar_visualizacoes(df, file_name):
    """
    Gera visualizações completas para um arquivo
    """
    numeric_cols = df.select_dtypes(include=[np.number]).columns
    
    if len(numeric_cols) == 0:
        print(f"Sem dados numéricos em {file_name}")
        return
    
    # Figura principal com 6 subplots
    fig = plt.figure(figsize=(20, 14))
    fig.suptitle(f'Análise Completa: {file_name}', fontsize=18, fontweight='bold', y=0.995)
    
    # 1. Série temporal das primeiras 3 colunas
    ax1 = plt.subplot(3, 3, 1)
    cols_to_plot = numeric_cols[:3]
    for col in cols_to_plot:
        ax1.plot(df[col], label=col, linewidth=1.5, alpha=0.7)
    ax1.set_title('Série Temporal (Primeiras 3 colunas)', fontweight='bold', fontsize=11)
    ax1.set_xlabel('Índice')
    ax1.set_ylabel('Valor')
    ax1.legend(fontsize=8)
    ax1.grid(True, alpha=0.3)
    
    # 2. Histograma da primeira coluna
    ax2 = plt.subplot(3, 3, 2)
    first_col = numeric_cols[0]
    ax2.hist(df[first_col].dropna(), bins=40, color='skyblue', edgecolor='black', alpha=0.7)
    ax2.set_title(f'Distribuição: {first_col}', fontweight='bold', fontsize=11)
    ax2.set_xlabel('Valor')
    ax2.set_ylabel('Frequência')
    ax2.grid(True, alpha=0.3, axis='y')
    
    # 3. Box plot
    ax3 = plt.subplot(3, 3, 3)
    data_to_box = [df[col].dropna() for col in cols_to_plot]
    bp = ax3.boxplot(data_to_box, labels=[c[:15] for c in cols_to_plot], patch_artist=True)
    for patch in bp['boxes']:
        patch.set_facecolor('lightblue')
    ax3.set_title('Box Plot', fontweight='bold', fontsize=11)
    ax3.set_ylabel('Valor')
    ax3.tick_params(axis='x', rotation=45, labelsize=8)
    ax3.grid(True, alpha=0.3, axis='y')
    
    # 4. Matriz de correlação
    ax4 = plt.subplot(3, 3, 4)
    if len(numeric_cols) > 1:
        corr = df[numeric_cols[:6]].corr()
        sns.heatmap(corr, annot=True, fmt='.2f', cmap='coolwarm', center=0, 
                    ax=ax4, cbar_kws={'label': 'Correlação'}, linewidths=0.5, 
                    annot_kws={'size': 8})
        ax4.set_title('Matriz de Correlação', fontweight='bold', fontsize=11)
    else:
        ax4.text(0.5, 0.5, 'Apenas 1 coluna', ha='center', va='center')
        ax4.set_xticks([])
        ax4.set_yticks([])
    
    # 5. Série temporal de todas as colunas numéricas (normalizada)
    ax5 = plt.subplot(3, 3, 5)
    for col in numeric_cols[:5]:
        normalized = (df[col] - df[col].min()) / (df[col].max() - df[col].min())
        ax5.plot(normalized, label=col[:15], linewidth=1, alpha=0.7)
    ax5.set_title('Série Temporal Normalizada (até 5 colunas)', fontweight='bold', fontsize=11)
    ax5.set_xlabel('Índice')
    ax5.set_ylabel('Valor Normalizado (0-1)')
    ax5.legend(fontsize=7)
    ax5.grid(True, alpha=0.3)
    
    # 6. Desvio padrão móvel
    ax6 = plt.subplot(3, 3, 6)
    rolling_std = df[first_col].rolling(window=max(5, len(df)//50)).std()
    ax6.plot(rolling_std, color='purple', linewidth=2)
    ax6.fill_between(range(len(rolling_std)), rolling_std, alpha=0.3, color='purple')
    ax6.set_title('Variabilidade (Desvio Padrão Móvel)', fontweight='bold', fontsize=11)
    ax6.set_xlabel('Índice')
    ax6.set_ylabel('Desvio Padrão')
    ax6.grid(True, alpha=0.3)
    
    # 7. Distribuição de todas colunas numéricas
    ax7 = plt.subplot(3, 3, 7)
    for col in numeric_cols[:4]:
        ax7.hist(df[col].dropna(), bins=20, alpha=0.5, label=col[:15])
    ax7.set_title('Distribuições Sobrepostas', fontweight='bold', fontsize=11)
    ax7.set_xlabel('Valor')
    ax7.set_ylabel('Frequência')
    ax7.legend(fontsize=7)
    ax7.grid(True, alpha=0.3, axis='y')
    
    # 8. Scatter plot entre primeiras 2 colunas
    ax8 = plt.subplot(3, 3, 8)
    if len(numeric_cols) >= 2:
        ax8.scatter(df[numeric_cols[0]], df[numeric_cols[1]], alpha=0.5, s=20)
        ax8.set_xlabel(numeric_cols[0][:20])
        ax8.set_ylabel(numeric_cols[1][:20])
        ax8.set_title(f'Scatter: {numeric_cols[0][:15]} vs {numeric_cols[1][:15]}', 
                     fontweight='bold', fontsize=11)
        ax8.grid(True, alpha=0.3)
    else:
        ax8.text(0.5, 0.5, 'Insuficiente dados', ha='center', va='center')
    
    # 9. Estatísticas em texto
    ax9 = plt.subplot(3, 3, 9)
    ax9.axis('off')
    
    stats_text = f"""
RESUMO ESTATÍSTICO

Dimensões: {df.shape[0]:,} x {df.shape[1]}
Numéricas: {len(numeric_cols)} de {df.shape[1]}

{first_col[:25]}:
  Média: {df[first_col].mean():.6f}
  Mediana: {df[first_col].median():.6f}
  Std: {df[first_col].std():.6f}
  Mín: {df[first_col].min():.6f}
  Máx: {df[first_col].max():.6f}
  
Faltantes: {df.isnull().sum().sum()}
Duplicatas: {df.duplicated().sum()}
Memória: {df.memory_usage(deep=True).sum()/1024:.1f} KB
"""
    
    ax9.text(0.05, 0.5, stats_text, fontsize=9, family='monospace',
             verticalalignment='center', bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.8))
    
    plt.tight_layout()
    plt.show()

# Exemplo de visualização
gerar_visualizacoes(df_example, os.path.basename(primeiro_arquivo))
```

## Analisar Todos os Arquivos (9ª a 18ª) - Loop Completo

```python
print(f"\n{'#'*100}")
print(f"{'ANALISANDO ARQUIVOS 9ª A 18ª POSIÇÃO':^100}")
print(f"{'#'*100}\n")

# Armazenar dados para análise posterior
dados_arquivos = {}

for idx, file_path in enumerate(selected_files, 9):
    file_name = os.path.basename(file_path)
    print(f"\n>>> Processando arquivo {idx}/18: {file_name}\n")
    
    df, ctx = gerar_relatorio_detalhado(file_path, idx)
    dados_arquivos[file_name] = {'df': df, 'contexto': ctx}
    
    # Gerar visualizações
    gerar_visualizacoes(df, file_name)
    
    print(f"\n✓ Arquivo {idx} processado com sucesso!")
    print(f"{'─'*100}\n")

print(f"\n{'#'*100}")
print(f"{'ANÁLISE DE TODOS OS ARQUIVOS CONCLUÍDA':^100}")
print(f"{'#'*100}")
```

## Detecção de Anomalias em Todos os Arquivos

```python
print("\n" + "="*120)
print(f"{'DETECÇÃO DE ANOMALIAS (Z-score > 3) - TODOS OS ARQUIVOS':^120}")
print("="*120 + "\n")

anomalias_resumo = []

for file_name, data in dados_arquivos.items():
    df = data['df']
    numeric_cols = df.select_dtypes(include=[np.number]).columns
    
    anomalias_por_coluna = {}
    
    for col in numeric_cols:
        try:
            z_scores = np.abs(zscore(df[col].dropna()))
            anomaly_count = np.sum(z_scores > 3)
            if anomaly_count > 0:
                pct = (anomaly_count / len(df)) * 100
                anomalias_por_coluna[col] = {'count': anomaly_count, 'pct': pct}
        except:
            pass
    
    if anomalias_por_coluna:
        print(f"\n📍 {file_name.replace('carbonZ_2018-07-18-12-10-11_no_ground_truth-', '')}:")
        for col, info in sorted(anomalias_por_coluna.items(), key=lambda x: x[1]['count'], reverse=True):
            print(f"   • {col:<35} {info['count']:>4} anomalias ({info['pct']:>5.2f}%)")
            anomalias_resumo.append({
                'Arquivo': file_name[:50],
                'Coluna': col,
                'Anomalias': info['count'],
                'Percentual': info['pct']
            })
    else:
        print(f"\n✅ {file_name.replace('carbonZ_2018-07-18-12-10-11_no_ground_truth-', '')}:")
        print(f"   • Nenhuma anomalia significativa detectada")

if anomalias_resumo:
    print(f"\n{'─'*120}")
    print(f"\nTOP 10 ANOMALIAS:\n")
    anomalias_df = pd.DataFrame(anomalias_resumo).sort_values('Anomalias', ascending=False).head(10)
    print(anomalias_df.to_string(index=False))

print(f"\n{'='*120}\n")
```
