## 📂 Carregando arquivos CSV

```python
files = [
    'carbonZ_2018-07-18-15-53-31_1_engine_failure-mavros-global_position-local.csv',
    'carbonZ_2018-07-18-15-53-31_1_engine_failure-mavros-global_position-raw-fix.csv',
    'carbonZ_2018-07-18-15-53-31_1_engine_failure-mavros-global_position-raw-gps_vel.csv',
    'carbonZ_2018-07-18-15-53-31_1_engine_failure-mavros-global_position-rel_alt.csv',
    'carbonZ_2018-07-18-15-53-31_1_engine_failure-mavros-imu-atm_pressure.csv',
    'carbonZ_2018-07-18-15-53-31_1_engine_failure-mavros-imu-data.csv',
    'carbonZ_2018-07-18-15-53-31_1_engine_failure-mavros-imu-data_raw.csv',
    'carbonZ_2018-07-18-15-53-31_1_engine_failure-mavros-imu-mag.csv',
    'carbonZ_2018-07-18-15-53-31_1_engine_failure-mavros-imu-temperature.csv',
    'carbonZ_2018-07-18-15-53-31_1_engine_failure-mavros-local_position-odom.csv',
]

dataframes = {}
for file in files:
    try:
        df = pd.read_csv(file)
        dataframes[file] = df
        print(f'✔ {file} carregado - shape:', df.shape)
    except Exception as e:
        print(f'❌ Erro ao carregar {file}:', e)
```

## 🔎 Visualizar primeiras linhas de cada arquivo

## 📈 Exemplo: Plot da Altitude Relativa

```python
rel_alt_file = [f for f in files if 'rel_alt' in f][0]
df_alt = dataframes.get(rel_alt_file)

if df_alt is not None and 'field.data' in df_alt.columns:
    df_alt['time'] = df_alt['%time'] - df_alt['%time'].iloc[0]
    df_alt['time'] = df_alt['time'] / 1e9  # converter ns para segundos

    plt.plot(df_alt['time'], df_alt['field.data'])
    plt.xlabel('Tempo (s)')
    plt.ylabel('Altitude Relativa (m)')
    plt.title('Altitude Relativa durante o voo')
    plt.show()
else:
    print('Coluna field.data não encontrada.')
```
